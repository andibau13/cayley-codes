```python
import numpy as np
from core import random_gens, hermitian_transpose, syzygies

n, m = 1, 2        # f: R^n -> R^m
l_init, w = 1, 5   # depth and weight of the random columns of f
l_max = 2          # star depth for the syzygy computation

rng = np.random.default_rng()
f = random_gens(m, n, l_init, w, rng)   # n random elements of R^m, as a map R^n -> R^m
print(f)

# annihilator of im(f) w.r.t. <x, y> = int(x^* y):  ker(f^dagger): R^k -> R^m
ann = syzygies(hermitian_transpose(f), l_max)
print(f"\nannihilator: {ann.n} generator(s), level counts {ann.level_counts}")
print(ann)
```
